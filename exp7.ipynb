```python
#Fibonacci series using while loop
n= int(input("Enter number of terms:"))
a=0
b=1
i=1
while i<=n:
  print(a ,end=" ")# end command used to keep it parallel
  c = a+b
  a= b
  b=c
  i +=1
```

```python
#limiting the series
limit= int(input("enter a limit:")) #term inside the series shouldnt have a greater value than the value entered
a=0
b=1
while a<= limit:
  print(a,end=" ")
  a,b=b, a+b
```

```python
# reverse a number
num=int(input("Enter a number: "))
rev =0
while num >0:
  digit = num %10
  rev = rev * 10 + digit
  num //=10
print("reverse num :",rev)
```

```python
# check palindrome number/string
num= int(input("enter a number:"))
temp = num
rev =0
while num >0:
  rev =rev * 10 + num % 10
  num//= 10
if temp ==rev:
  print("Palindrome")
else:
  print("Not a Palindrome")
```

```python

# check palindrome for string
s= str(input("enter a string: "))# only works in the same case
i,j=0,len(s)-1
is_palindrome = True
while i<j:
  if s [i] != s[j]:
   is_palindrome = False
  break
  i += 1
  j -=1
if is_palindrome:
  print("palindrome")
else:
  print("not palindrome")


```

```python
#check palindrome  without while loop only if loop
st=input ("enter input: ")
rev = st[::-1]
if st==rev:
  print("palindrome")
else:
  print("not palindrome")
```

```python
#count digits in a number
num=int(input("Enter a number: "))
count = 0
while num >0:
  count +=1
  num //=10
print("number of digits =",count)
#m2 use len
```

Break Statement:
With break statement we can stop the loop even if the while condition is true

```python
#exit the loop when i is 3:
i=1
while i<6:
  print(i)
  if i ==3:
    break
  i +=1
```

```python
#search a element in a list
nums=[10,20,30,40,50]
key= int(input("enter element to search: "))
i=0
while i< len(nums):
  if nums[i]== key:
    print("element found at index",i)
    break
  i += 1
else:
  print("element not found")  #putting while with else
```
